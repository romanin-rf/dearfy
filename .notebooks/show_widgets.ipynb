```python
import inspect
from typing_extensions import Any, Iterable, Iterator, TypeVar, Callable


T = TypeVar('T')


def blind(*iters: Iterable[T]) -> Iterator[T]:
    yield from ()
    for it in iters:
        yield from it


def search_unique_args(*methods: Callable[..., Any]) -> list[str]:
    methods_annotations = {
        method.__qualname__: method.__annotations__ 
        for method in methods 
            if inspect.isfunction(method)
    }
    result, handlered = [], ['return']
    for attr in blind(*[list(_.keys()) for _ in methods_annotations.values()]):
        if attr in handlered:
            continue
        count = 0
        for method_annotation in methods_annotations.values():
            if attr in list(method_annotation.keys()):
                count += 1
                continue
        if count == len(list(methods_annotations.keys())):
            result.append(attr)
        handlered.append(attr)
    return result
```

```python
import dearpygui.dearpygui as dpg

uniques_group = {
    'Item': search_unique_args(dpg.add_text, dpg.add_button, dpg.add_image, dpg.add_combo),
    'ItemContainer': search_unique_args(dpg.add_group, dpg.add_table),
    'Container': search_unique_args(dpg.add_window, dpg.add_tab),
}

console.print(uniques_group)
```
