```python
from rich.console import Console

console = Console()
```

```python
import inspect
import dearpygui.dearpygui as dpg
from types import FunctionType

methods: list[FunctionType] = [
    dpg.add_button,
    dpg.add_text,
    dpg.add_image,
    dpg.window,
    dpg.group,
    dpg.table
]

methods_annotations: dict[str, dict[str, object]] = {
    method.__qualname__: method.__annotations__ for method in methods if inspect.isfunction(method)
}
```

```python
unique_methods_annotations = []

for _ in methods_annotations.values():
    unique_methods_annotations += list(_.keys())

unique_methods_annotations: list[str] = list(set(unique_methods_annotations))

del _
```

```python
general_methods_annotations = []

for unique_method_annotation in unique_methods_annotations:
    count = 0
    for method_annotation in methods_annotations.values():
        if unique_method_annotation in list(method_annotation.keys()):
            count += 1
            continue
    if count == len(list(methods_annotations.keys())):
        general_methods_annotations.append(unique_method_annotation)

general_methods_annotations.remove('return')

del unique_methods_annotations, method_annotation
```

```python
def sort_method_annotation(value: str) -> int:
    names = list(methods[0].__annotations__.keys())
    for index, name in enumerate(names):
        if value == name:
            return index, name
    raise ValueError(repr(value))

general_methods_annotations.sort(key=sort_method_annotation)

console.print(general_methods_annotations)
```
