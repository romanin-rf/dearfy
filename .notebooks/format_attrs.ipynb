```python
import re
from typing_extensions import Iterator, TypedDict

STRIP_CHARS = ' \r\n\'\":;,'

REGEX_FULL = r'\s*(?P<name>.*)\s*:\s*(?P<type>.*)\s*=\s*(?P<defaut>.*)\s*(?:,|)'

class Attr(TypedDict):
    name: str
    type: str | None
    default: str | None

attrs = []
for line in text.split('\n'):
    attr = line.split(':')[0].strip()
    if attr:
        attrs.append(attr)

def parse_attrs(text: str) -> Iterator[Attr]:
    yield from ()
    for line in text.split('\n'):
        if (result := re.search(REGEX_FULL, line)) is not None:
            result
        yield 

console.print(attrs)
```
